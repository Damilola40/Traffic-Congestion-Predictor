```python
# Data Cleaning: Encoding

# Encode days in their chronological order (Monday=0, Sunday=6).
day_encoding = {'Monday': 0, 'Tuesday': 1, 'Wednesday': 2, 'Thursday': 3,
                'Friday': 4, 'Saturday': 5, 'Sunday': 6}
traffic_df_copy['day_encoded'] = traffic_df_copy['day'].map(day_encoding).astype('Int64')
print(f"")

# Encode traffic-light patterns. G=green, Y=yellow, and R=red.
traffic_pattern_encoding = {'G': 0, 'Y': 1, 'R': 2, 'G/Y': 3, 'G/R': 4,
                            'Y/R': 5, 'G/Y/R': 6}
traffic_df_copy['traffic_pattern_encoded'] = (
    traffic_df_copy['traffic_pattern'].map(traffic_pattern_encoding).astype('Int64')
)

# Encode traffic_level. low, medium, high
traffic_level_encoding = {'low': 0, 'medium': 1, 'high': 2}
traffic_df_copy['traffic_level_encoded'] = (
    traffic_df_copy['traffic_level'].map(traffic_level_encoding).astype('Int64')
)

# Encode congestion location.
congestion_location_encoding = {'none evident': 0,
                                'road segment': 1,
                                'junction': 2,
                                'mixed': 3}
traffic_df_copy['congestion_location_encoded'] = (
    traffic_df_copy['congestion_location'].map(congestion_location_encoding).astype('Int64')
)

print(f"The columns: day, traffic_pattern, traffic_level and congestion_location encoded successfully !!!")
```
